## 2. Data Cleaning

Handle missing values, fix types, remove duplicates, etc.

```python
df = df_raw.copy()

# Remove rows with unreliable estimate flags
# Keep only rows where FLAG is NaN (no flag = reliable)
print(f"Before flag filter: {len(df)} rows")
df = df[df['FLAG'].isna()]
print(f"After flag filter: {len(df)} rows")

# Remove rows where ESTIMATE is missing
print(f"Before estimate filter: {len(df)} rows")
df = df[df['ESTIMATE'].notna()]
print(f"After estimate filter: {len(df)} rows")

# Drop columns that are entirely or mostly null and not useful
cols_to_drop = ['SUBTOPIC', 'SUBTOPIC_ID', 'NESTING_LABEL', 'NESTING_LABEL_ID',
                'TIME_PERIOD_ID', 'STANDARD_ERROR', 'FLAG', 'FOOTNOTE_ID_LIST']
df = df.drop(columns=cols_to_drop)

print(f"\nFinal shape: {df.shape}")
print(f"Columns kept: {list(df.columns)}")
df.head()
```

```python
# Year-over-year change for each topic + subgroup combination
df = df.sort_values(['TOPIC', 'GROUP', 'SUBGROUP', 'TIME_PERIOD'])
df['YOY_CHANGE'] = df.groupby(['TOPIC', 'GROUP', 'SUBGROUP'])['ESTIMATE'].diff()

# Confidence interval width (narrower = more reliable estimate)
df['CI_WIDTH'] = df['ESTIMATE_UCI'] - df['ESTIMATE_LCI']

print(df[['TOPIC', 'SUBGROUP', 'TIME_PERIOD', 'ESTIMATE', 'YOY_CHANGE', 'CI_WIDTH']].head(12))
```

```python
# Topics related to access barriers
access_topics = [
    'Delayed getting medical care due to cost among adults',
    'Did not get needed medical care due to cost',
    'Did not get needed mental health care due to cost',
    'Did not take medication as prescribed to save money',
    'Uninsured at time of interview among adults',
    'Uninsured for at least part of the past year among adults',
    'Has a usual place of care among adults',
    'Fair or poor health status in adults'
]

df_access = df[df['TOPIC'].isin(access_topics)]
print(f"Access-related rows: {len(df_access)}")
print(f"Topics included:\n{df_access['TOPIC'].value_counts()}")
```

```python
# Create a reusable function so we can crank these out fast
def plot_disparity(df, topic, group):
    max_year = df['TIME_PERIOD'].max()
    plot_data = df[
        (df['TOPIC'] == topic) &
        (df['GROUP'] == group) &
        (df['TIME_PERIOD'] == max_year)
    ].sort_values('ESTIMATE', ascending=True)

    if len(plot_data) == 0:
        print(f"No data for {topic} x {group}")
        return

    fig = px.bar(
        plot_data,
        x='ESTIMATE',
        y='SUBGROUP',
        orientation='h',
        title=f'{topic}<br>by {group} ({max_year})',
        labels={'ESTIMATE': 'Percent of Population', 'SUBGROUP': ''},
        error_x_minus=plot_data['ESTIMATE'] - plot_data['ESTIMATE_LCI'],
        error_x=plot_data['ESTIMATE_UCI'] - plot_data['ESTIMATE'],
        color='ESTIMATE',
        color_continuous_scale='Reds'
    )
    fig.update_layout(showlegend=False, height=400)
    fig.show()

# Now let's see several angles
plot_disparity(df_access, 'Did not get needed medical care due to cost', 'Race and Hispanic origin')
plot_disparity(df_access, 'Did not get needed medical care due to cost', 'Education')
plot_disparity(df_access, 'Did not get needed mental health care due to cost', 'Poverty level')
plot_disparity(df_access, 'Uninsured at time of interview among adults', 'Race and Hispanic origin')
```

```python
# Dumbbell: Compare lowest vs highest poverty level across access topics
import plotly.graph_objects as go

max_year = df_access['TIME_PERIOD'].max()
topics_short = [
    'Did not get needed medical care due to cost',
    'Did not get needed mental health care due to cost',
    'Delayed getting medical care due to cost among adults',
    'Did not take medication as prescribed to save money'
]

fig = go.Figure()

for topic in topics_short:
    low = df_access[(df_access['TOPIC'] == topic) &
                     (df_access['SUBGROUP'] == '<100% FPL') &
                     (df_access['TIME_PERIOD'] == max_year)]['ESTIMATE'].values
    high = df_access[(df_access['TOPIC'] == topic) &
                      (df_access['SUBGROUP'] == '≥200% FPL') &
                      (df_access['TIME_PERIOD'] == max_year)]['ESTIMATE'].values

    if len(low) > 0 and len(high) > 0:
        # Line connecting the two dots
        fig.add_trace(go.Scatter(
            x=[high[0], low[0]],
            y=[topic, topic],
            mode='lines',
            line=dict(color='gray', width=3),
            showlegend=False
        ))
        # High income dot
        fig.add_trace(go.Scatter(
            x=[high[0]], y=[topic],
            mode='markers',
            marker=dict(size=14, color='#4CAF50'),
            name='≥200% FPL',
            showlegend=(topic == topics_short[0])
        ))
        # Low income dot
        fig.add_trace(go.Scatter(
            x=[low[0]], y=[topic],
            mode='markers',
            marker=dict(size=14, color='#F44336'),
            name='<100% FPL',
            showlegend=(topic == topics_short[0])
        ))

fig.update_layout(
    title='The Poverty Gap in Healthcare Access (2024)<br><sup>Distance between dots = size of the disparity</sup>',
    xaxis_title='Percent of Population',
    height=450,
    margin=dict(l=20),
    yaxis=dict(autorange='reversed')
)
fig.show()
```

```python
# Overall trends (Total population) for key access topics
covid_topics = [
    'Delayed getting medical care due to cost among adults',
    'Did not get needed medical care due to cost',
    'Did not get needed mental health care due to cost',
    'Uninsured at time of interview among adults'
]

overall = df_access[
    (df_access['TOPIC'].isin(covid_topics)) &
    (df_access['GROUP'] == 'Total')
].sort_values('TIME_PERIOD')

fig = px.line(
    overall,
    x='TIME_PERIOD',
    y='ESTIMATE',
    color='TOPIC',
    markers=True,
    title='Healthcare Access Barriers: The Big Picture (2019-2024)<br><sup>Total adult population</sup>',
    labels={'ESTIMATE': '% of Population', 'TIME_PERIOD': 'Year', 'TOPIC': ''}
)
fig.update_layout(height=500, plot_bgcolor='rgba(0,0,0,0)', legend=dict(y=-0.3, orientation='h'))
fig.show()
```

```python
# Gap analysis: difference between worst-off and best-off subgroup per year
topic = 'Did not get needed medical care due to cost'

gap_groups = ['Poverty level', 'Race and Hispanic origin', 'Education']

fig = go.Figure()
colors = {'Poverty level': '#e63946', 'Race and Hispanic origin': '#457b9d', 'Education': '#2a9d8f'}

for group in gap_groups:
    group_data = df_access[
        (df_access['TOPIC'] == topic) &
        (df_access['GROUP'] == group)
    ]

    # For each year, get max and min estimate
    gap_by_year = group_data.groupby('TIME_PERIOD').agg(
        worst=('ESTIMATE', 'max'),
        best=('ESTIMATE', 'min')
    )
    gap_by_year['gap'] = gap_by_year['worst'] - gap_by_year['best']

    fig.add_trace(go.Scatter(
        x=gap_by_year.index,
        y=gap_by_year['gap'],
        mode='lines+markers+text',
        name=group,
        line=dict(width=3, color=colors[group]),
        marker=dict(size=10),
        text=[f'{v:.1f}pp' for v in gap_by_year['gap']],
        textposition='top center'
    ))

fig.update_layout(
    title='Did COVID Widen the Gap?<br><sup>Disparity gap (worst − best subgroup) in missed medical care due to cost</sup>',
    xaxis_title='Year',
    yaxis_title='Gap (percentage points)',
    height=500,
    plot_bgcolor='rgba(0,0,0,0)'
)
fig.show()
```

```python
# Gap widening across all four topics, using poverty level
fig = go.Figure()
colors_list = ['#e63946', '#457b9d', '#2a9d8f', '#e9c46a']

for i, topic in enumerate(covid_topics):
    group_data = df_access[
        (df_access['TOPIC'] == topic) &
        (df_access['GROUP'] == 'Poverty level')
    ]

    gap_by_year = group_data.groupby('TIME_PERIOD').agg(
        worst=('ESTIMATE', 'max'),
        best=('ESTIMATE', 'min')
    )
    gap_by_year['gap'] = gap_by_year['worst'] - gap_by_year['best']

    fig.add_trace(go.Scatter(
        x=gap_by_year.index,
        y=gap_by_year['gap'],
        mode='lines+markers',
        name=topic[:40] + '...' if len(topic) > 40 else topic,
        line=dict(width=3, color=colors_list[i]),
        marker=dict(size=10)
    ))

fig.update_layout(
    title='Which Barriers Show the Widest Poverty Gap?<br><sup>Gap between <100% FPL and ≥200% FPL over time</sup>',
    xaxis_title='Year',
    yaxis_title='Gap (percentage points)',
    height=500,
    plot_bgcolor='rgba(0,0,0,0)',
    legend=dict(y=-0.3, orientation='h')
)
fig.show()
```

```python
# Treemap: Biggest disparity gaps across all topics (poverty level only)
max_year = df['TIME_PERIOD'].max()

tree_data = []
for topic in df['TOPIC'].unique():
    group_data = df[
        (df['TOPIC'] == topic) &
        (df['GROUP'] == 'Poverty level') &
        (df['TIME_PERIOD'] == max_year)
    ]
    if len(group_data) >= 2:
        gap = group_data['ESTIMATE'].max() - group_data['ESTIMATE'].min()
        worst = group_data.loc[group_data['ESTIMATE'].idxmax(), 'SUBGROUP']
        if gap > 0:
            tree_data.append({
                'Topic': topic,
                'Gap': round(gap, 1),
                'Worst Subgroup': worst,
                'Category': group_data.iloc[0]['TAXONOMY']
            })

tree_df = pd.DataFrame(tree_data)

fig = px.treemap(
    tree_df,
    path=['Category', 'Topic'],
    values='Gap',
    color='Gap',
    color_continuous_scale='YlOrRd',
    title='Poverty Disparity Gaps by Health Topic (2024)<br><sup>Larger block = bigger gap between poorest and wealthiest</sup>',
    hover_data=['Worst Subgroup', 'Gap']
)
fig.update_layout(height=600)
fig.update_traces(textinfo='label+value', texttemplate='%{label}<br>%{value}pp')
fig.show()
```

```python
# ===== FETCH CENSUS ACS COUNTY-LEVEL UNINSURED RATES + DEMOGRAPHICS =====
import requests
import pandas as pd
import json

print("Fetching Census ACS county-level data (uninsured + demographics)...")

# Census ACS API endpoint (2022 data)
# Variables:
# S2701_C05_001E = Uninsured rate (%)
# S1701_C03_001E = Poverty rate (%)

api_url = "https://api.census.gov/data/2022/acs/acs1/subject"
params = {
    "get": "NAME,S2701_C05_001E",
    "for": "county:*"
}

try:
    # Fetch uninsured rates
    print("  Fetching uninsured rates...")
    response = requests.get(api_url, params=params, timeout=30)
    response.raise_for_status()
    data = response.json()

    headers = data[0]
    rows = data[1:]
    print(f"  ✅ Fetched {len(rows):,} county records")

    # Create DataFrame
    df_counties = pd.DataFrame(rows, columns=headers)
    df_counties.columns = ["county_name", "uninsured_pct", "state_fips", "county_fips"]
    df_counties["fips"] = df_counties["state_fips"] + df_counties["county_fips"]
    df_counties["uninsured_pct"] = pd.to_numeric(df_counties["uninsured_pct"], errors="coerce")
    df_counties = df_counties.dropna(subset=["uninsured_pct"])

    # Fetch demographic data (poverty rates)
    print("  Fetching poverty rates...")
    demo_params = {
        "get": "NAME,S1701_C03_001E",
        "for": "county:*"
    }
    demo_response = requests.get(api_url, params=demo_params, timeout=30)
    demo_data = demo_response.json()
    df_poverty = pd.DataFrame(demo_data[1:], columns=demo_data[0])
    df_poverty.columns = ["county_name", "poverty_pct", "state_fips", "county_fips"]
    df_poverty["fips"] = df_poverty["state_fips"] + df_poverty["county_fips"]
    df_poverty["poverty_pct"] = pd.to_numeric(df_poverty["poverty_pct"], errors="coerce")
    print(f"  ✅ Fetched {len(df_poverty)} poverty records")

    # Merge data
    df_counties = df_counties.merge(df_poverty[["fips", "poverty_pct"]], on="fips", how="left")

    print("\nData summary:")
    print(f"  Counties: {len(df_counties)}")
    print(f"  Uninsured rate range: {df_counties['uninsured_pct'].min():.1f}% - {df_counties['uninsured_pct'].max():.1f}%")
    print(f"  Mean uninsured: {df_counties['uninsured_pct'].mean():.1f}%")
    print(f"  Mean poverty: {df_counties['poverty_pct'].mean():.1f}%")

    print("\nHighest uninsured rates:")
    print(df_counties.nlargest(5, "uninsured_pct")[["county_name", "uninsured_pct", "poverty_pct", "fips"]])

except Exception as e:
    print(f"❌ Error fetching data: {e}")
    df_counties = None
```

```python
# ===== EXPORT COUNTY MAP DATA WITH DEMOGRAPHICS TO JSON =====
import os

if df_counties is not None:
    print("Preparing data for county map with demographics...")

    # Create map-ready format with demographics
    map_data = []
    for _, row in df_counties.iterrows():
        county_dict = {
            "fips": row["fips"],
            "county_name": row["county_name"],
            "uninsured_pct": float(row["uninsured_pct"])
        }

        # Add poverty rate if available
        if pd.notna(row.get("poverty_pct")):
            county_dict["poverty_pct"] = float(row["poverty_pct"])

        map_data.append(county_dict)

    # Create output directory if needed
    output_dir = "/content/drive/MyDrive/datathon-2026/data"
    os.makedirs(output_dir, exist_ok=True)

    # Export to JSON
    output_path = os.path.join(output_dir, "county_map_data.json")
    with open(output_path, "w") as f:
        json.dump(map_data, f, indent=2)

    print(f"✅ Exported {len(map_data)} counties to {output_path}")
    print("\nSample output:")
    print(json.dumps(map_data[:3], indent=2))
else:
    print("❌ No data available to export")
```
